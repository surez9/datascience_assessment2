```python
from scipy import stats

players = pd.read_csv("players.csv").sort_values("snapshot").drop_duplicates("player_id", keep="last")
incidents = pd.read_csv("post_match_incidents.csv")
stats_df = pd.read_csv("post_match_stats.csv")
details = pd.read_csv("post_match_details.csv")
```

```python
# Build a team_id -> team name lookup from home and away columns combined
home_map = details[["home_team_id", "home_team"]].rename(columns={"home_team_id": "team_id", "home_team": "team_name"})
away_map = details[["away_team_id", "away_team"]].rename(columns={"away_team_id": "team_id", "away_team": "team_name"})
team_names = pd.concat([home_map, away_map]).drop_duplicates("team_id")

cards = incidents[incidents["type"] == "card"]
yellow = cards[cards["card_type"].isin(["yellow", "yellowRed"])]
red = cards[cards["card_type"].isin(["red", "yellowRed"])]

merged = players[["player_id", "position", "age", "team_id", "caps"]].merge(
    yellow.groupby("player_id").size().rename("yellow"), on="player_id", how="left"
).merge(
    red.groupby("player_id").size().rename("red"), on="player_id", how="left"
).fillna(0)
```

```python
# players.csv has multiple snapshots per player. Keep only the latest one.
players_clean = players.sort_values("snapshot").drop_duplicates(subset="player_id", keep="last")

# Filter to yellow cards only
yellow_cards = incidents[incidents["card_type"] == "yellow"]
print("Total yellow cards:", len(yellow_cards))

```

```python
# Count yellow cards per player
card_counts = yellow_cards.groupby("player_id").size().reset_index(name="yellow_cards")

# Merge onto full player list, so players with zero cards are included
player_cards = players_clean[["player_id", "name", "position"]].merge(
    card_counts, on="player_id", how="left"
)
player_cards["yellow_cards"] = player_cards["yellow_cards"].fillna(0)

# Split into defenders (DF) and midfielders (MF)
defenders = player_cards[player_cards["position"] == "DF"]
midfielders = player_cards[player_cards["position"] == "MF"]

def_mean = defenders["yellow_cards"].mean()
mid_mean = midfielders["yellow_cards"].mean()

print(f"\nDefenders: n={len(defenders)}, total={int(defenders['yellow_cards'].sum())}, mean={def_mean:.4f}")
print(f"Midfielders: n={len(midfielders)}, total={int(midfielders['yellow_cards'].sum())}, mean={mid_mean:.4f}")

```

```python
# Q significance: forwards vs defenders

players = pd.read_csv("players.csv").sort_values("snapshot").drop_duplicates("player_id", keep="last")
incidents = pd.read_csv("post_match_incidents.csv")
stats_df = pd.read_csv("post_match_stats.csv")
 
cards = incidents[incidents["type"] == "card"]
yellow_cards = incidents[incidents["card_type"].isin(["yellow", "yellowRed"])]
red = cards[cards["card_type"].isin(["red", "yellowRed"])]
 
pc = players[["player_id", "name", "position", "age"]].merge(
    yellow.groupby("player_id").size().rename("yellow"), on="player_id", how="left"
).merge(
    red.groupby("player_id").size().rename("red"), on="player_id", how="left"
).fillna(0)
```

```python
#  yellow cards by position
print("Yellow cards by position:")
print(pc.groupby("position")["yellow"].agg(["count", "sum", "mean"]).round(2))
```

Forwards averaged 0.15 yellow cards per player (46 cards across 311 players). Defenders averaged 0.24 (104 cards across 426 players). Forwards got fewer cards.

```python
# Question: Red cards by position
print(pc.groupby("position")["red"].agg(["count", "sum"]))
```

Defenders picked up 7 red cards, midfielders 6, forwards 2, goalkeepers 0. Total across the tournament: 15 red cards. 
    This is too small a sample to draw a strong conclusion from.

```python
# Question: When in the match do most cards happen?
bins = pd.cut(cards["minute"], [-10, 45, 90, 130], labels=["1st half", "2nd half", "extra time"])
print(bins.value_counts())
```

168 cards happened in the second half, 110 in the first half, and 13 in extra time. Total: 291 cards. 
    The second half produces more cards than the first, likely from fatigue and rising match stakes late in games.

```python
# Questions: fouls vs cards
fd = stats_df[["home_fouls", "home_yellow_cards"]].dropna()
r5, p5 = stats.pearsonr(fd.home_fouls, fd.home_yellow_cards)
print(f"\nQ5  r={r5:.3f}  p={p5:.4f}")
```

```python
# Question: Which national team gets the most yellow cards?
merged = players[["player_id", "position", "age", "team_id", "caps"]].merge(
    yellow.groupby("player_id").size().rename("yellow"), on="player_id", how="left"
).merge(
    red.groupby("player_id").size().rename("red"), on="player_id", how="left"
).fillna(0)
```

```python
# cards by team
by_team = merged.groupby("team_id")[["yellow", "red"]].sum().merge(team_names, on="team_id", how="left")
print("\nQ most carded:\n", by_team.sort_values("yellow", ascending=False).head(10)[["team_name", "yellow", "red"]])
print("Q most disciplined:\n", by_team.sort_values("yellow").head(10)[["team_name", "yellow", "red"]])
```
